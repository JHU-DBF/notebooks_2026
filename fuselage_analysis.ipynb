```python
import numpy as np
```

```python
def calculate_twist_circular(outer_diameter, wall_thickness, length, G, torque):
    """
    Calculates the angle of twist for a hollow circular tube.

    Args:
        outer_diameter (float): Outer diameter of the tube (meters).
        wall_thickness (float): Wall thickness of the tube (meters).
        length (float): Length of the tube (meters).
        G (float): Shear modulus of the material (Pascals).
        torque (float): Applied torque (Newton-meters).

    Returns:
        tuple: Angle of twist in radians and degrees.
    """
    # Calculate radii
    r_outer = outer_diameter / 2
    r_inner = r_outer - wall_thickness

    # Calculate Polar Moment of Inertia (J)
    J = (np.pi / 2) * (r_outer**4 - r_inner**4)

    # Calculate angle of twist (theta)
    theta_rad = (torque * length) / (G * J)
    theta_deg = np.rad2deg(theta_rad)

    return theta_rad, theta_deg
```

```python
def calculate_twist_square(outer_side, wall_thickness, length, G, torque):
    """
    Calculates the angle of twist for a hollow square tube.

    Args:
        outer_side (float): Outer side length of the tube (meters).
        wall_thickness (float): Wall thickness of the tube (meters).
        length (float): Length of the tube (meters).
        G (float): Shear modulus of the material (Pascals).
        torque (float): Applied torque (Newton-meters).

    Returns:
        tuple: Angle of twist in radians and degrees.
    """
    # Calculate Mean Area (Am)
    Am = (outer_side - wall_thickness)**2

    # Calculate angle of twist (theta) using Bredt's theory
    theta_rad = (torque * length) / (4 * Am**2 * G * wall_thickness)
    theta_deg = np.rad2deg(theta_rad)

    return theta_rad, theta_deg
```

```python
inch_to_meter = 0.0254

# Material Properties (Carbon Fiber)
SHEAR_MODULUS_G = 5e9  # Pascals (5 GPa)

# Load
APPLIED_TORQUE = 10  # Newton-meters
```

```python
# --- Input Parameters (SI Units) ---

# Geometry
TUBE_LENGTH = 5 * 12 * inch_to_meter / 3  # meters
WALL_THICKNESS = 0.01 * 3 * inch_to_meter  # meters

SQUARE_OUTER_SIDE = 4.5 * inch_to_meter / 3
CIRCULAR_OUTER_DIAMETER = SQUARE_OUTER_SIDE

# --- Calculations ---
# Circular Tube
circ_rad, circ_deg_scaled = calculate_twist_circular(
    CIRCULAR_OUTER_DIAMETER, WALL_THICKNESS, TUBE_LENGTH, SHEAR_MODULUS_G, APPLIED_TORQUE
)

# Square Tube
sq_rad, sq_deg_scaled = calculate_twist_square(
    SQUARE_OUTER_SIDE, WALL_THICKNESS, TUBE_LENGTH, SHEAR_MODULUS_G, APPLIED_TORQUE
)

# --- Print Results ---
print("--- Torsion Calculation Results for SCALED DOWN BY 1/3 ---")
print(f"Applied Torque: {APPLIED_TORQUE} N·m\n")

print("Hollow Circular Tube:")
print(f"  Outer Diameter: {(CIRCULAR_OUTER_DIAMETER / inch_to_meter):.1f} inch")
print(f"  Angle of Twist: {circ_deg_scaled:.5f} degrees")
print("-" * 35)

print("Hollow Square Tube:")
print(f"  Outer Side: {(SQUARE_OUTER_SIDE / inch_to_meter):.1f} inch")
print(f"  Angle of Twist: {sq_deg_scaled:.5f} degrees")
print("-" * 35)
```

```python
# --- Input Parameters (SI Units) ---

# Geometry
TUBE_LENGTH = 5 * 12 * inch_to_meter  # meters
WALL_THICKNESS = 0.01 * 3 * inch_to_meter  # meters

SQUARE_OUTER_SIDE = 4.5 * inch_to_meter
CIRCULAR_OUTER_DIAMETER = SQUARE_OUTER_SIDE

# --- Calculations ---
# Circular Tube
circ_rad, circ_deg_full = calculate_twist_circular(
    CIRCULAR_OUTER_DIAMETER, WALL_THICKNESS, TUBE_LENGTH, SHEAR_MODULUS_G, APPLIED_TORQUE
)

# Square Tube
sq_rad, sq_deg_full = calculate_twist_square(
    SQUARE_OUTER_SIDE, WALL_THICKNESS, TUBE_LENGTH, SHEAR_MODULUS_G, APPLIED_TORQUE
)

# --- Print Results ---
print("--- Torsion Calculation Results ---")
print(f"Applied Torque: {APPLIED_TORQUE} N·m\n")

print("Hollow Circular Tube:")
print(f"  Outer Diameter: {(CIRCULAR_OUTER_DIAMETER / inch_to_meter):.1f} inch")
print(f"  Angle of Twist: {circ_deg_full:.5f} degrees")
print("-" * 35)

print("Hollow Square Tube:")
print(f"  Outer Side: {(SQUARE_OUTER_SIDE / inch_to_meter):.1f} inch")
print(f"  Angle of Twist: {sq_deg_full:.5f} degrees")
print("-" * 35)
```

```python
# --- Key Takeaways ---
print("================= KEY TAKEAWAYS ==================")
print("Analysis of the results shows two main points:\n")

# 1. Effect of Shape (for both scales)
print("1. EFFECT OF SHAPE: The circular cross-section is more efficient at resisting torsion.")
if sq_deg_full > circ_deg_full:
    shape_ratio_full = sq_deg_full / circ_deg_full
    print(f"   -> At FULL size, the circular tube is {shape_ratio_full:.2f} times stiffer (twists less) than the square tube.")

if sq_deg_scaled > circ_deg_scaled:
    shape_ratio_scaled = sq_deg_scaled / circ_deg_scaled
    print(f"   -> At 1/3 scale, the circular tube is {shape_ratio_scaled:.2f} times stiffer than the square tube.\n")


# 2. Effect of Scaling (for both shapes)
print("2. EFFECT OF SCALING: Torsional stiffness is highly sensitive to size.")
if circ_deg_scaled > circ_deg_full:
    scaling_ratio_circ = circ_deg_scaled / circ_deg_full
    print(f"   -> Scaling the CIRCULAR tube down to 1/3 size makes it twist {scaling_ratio_circ:.1f} times MORE.")

if sq_deg_scaled > sq_deg_full:
    scaling_ratio_sq = sq_deg_scaled / sq_deg_full
    print(f"   -> Scaling the SQUARE tube down to 1/3 size makes it twist {scaling_ratio_sq:.1f} times MORE.")
```
